```python
# ==========================================
# 实验1：基本运算、数据类型与字符串
# ==========================================
print("="*20, "实验1", "="*20)

# 1. 基本运算符
print("\n--- 任务1: 基本运算符 ---")
print(f"8 / 3 = {8 / 3}, 类型: {type(8 / 3)}")   # 除法，结果为浮点数
print(f"8 // 3 = {8 // 3}, 类型: {type(8 // 3)}") # 整除，结果为整数
print(f"8 % 3 = {8 % 3}, 类型: {type(8 % 3)}")   # 取模（求余数）
print(f"2 ** 7 = {2 ** 7}, 类型: {type(2 ** 7)}") # 幂运算

# 2. 数据类型检查
print("\n--- 任务2: 数据类型检查 ---")
int_val = 10
float_val = 3.14
str_val = "Hello"
bool_val = True
for val in [int_val, float_val, str_val, bool_val]:
    print(f"值: {val}, 类型: {type(val)}")

# 3. 类型转换与异常捕获
print("\n--- 任务3: 类型转换与异常捕获 ---")
print(f"int('123') = {int('123')}")
print(f"float('3.14') = {float('3.14')}")
print(f"int(3.9) = {int(3.9)} (向下取整)")
try:
    int("goat")
except ValueError as e:
    print(f"捕获到异常: {e}。原因：字符串 'goat' 不是合法的整数格式。")

# 4. 字符串操作
print("\n--- 任务4: 字符串操作 ---")
s = " Python Data Analysis "
print(f"原始字符串: '{s}'")
print(f"去除首尾空白: '{s.strip()}'")
print(f"转大写: '{s.upper()}'")
print(f"切片(前6个字符): '{s[:6]}'")
words = s.split()
print(f"单词拆分: {words}")
print(f"join连接: {'-'.join(words)}")

# 5. 字符串拼接、重复与不可变性
print("\n--- 任务5: 字符串拼接、重复与不可变性 ---")
str1 = "Hello"
str2 = "World"
print(f"拼接: {str1 + ' ' + str2}")
print(f"重复: {str1 * 3}")
try:
    str1[0] = 'h'
except TypeError as e:
    print(f"捕获到异常: {e}。原因：字符串是不可变类型，不能直接修改单个字符。")
new_str = 'h' + str1[1:]
print(f"构造新字符串: {new_str}")
```

```python
# ==========================================
# 实验2：函数、条件、循环与递归
# ==========================================
print("="*20, "实验2", "="*20)

# 1. score_level 函数
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0到100之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

print("\n--- 任务1: score_level ---")
print(f"score_level(95) = {score_level(95)}")
print(f"score_level(75) = {score_level(75)}")
try:
    score_level(105)
except ValueError as e:
    print(f"捕获到异常: {e}")

# 2. for 和 while 循环求和
print("\n--- 任务2: 循环求和 ---")
sum_for = 0
for i in range(1, 101):
    sum_for += i

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1

print(f"for循环求和: {sum_for}")
print(f"while循环求和: {sum_while}")
print(f"结果是否相同: {sum_for == sum_while}")

# 3. 递归版和迭代版阶乘
def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print("\n--- 任务3: 阶乘 ---")
for n in [0, 1, 5, 10]:
    print(f"n={n}: 递归={factorial_recursive(n)}, 迭代={factorial_iterative(n)}")

# 4. summary 函数
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    return min(values), max(values), sum(values) / len(values)

print("\n--- 任务4: summary ---")
data = [10, 20, 30, 40, 50]
min_val, max_val, avg_val = summary(data)
print(f"最小值: {min_val}, 最大值: {max_val}, 平均值: {avg_val}")

# 5. sum_all 函数
def sum_all(*args):
    return sum(args)

print("\n--- 任务5: sum_all ---")
print(f"无参数: {sum_all()}")
print(f"单个参数: {sum_all(5)}")
print(f"多个参数: {sum_all(1, 2, 3, 4, 5)}")
```

```python
# ==========================================
# 实验3：列表、字典、元组与基础统计
# ==========================================
print("="*20, "实验3", "="*20)

names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# 1. zip() 构建字典并遍历
print("\n--- 任务1: zip构建字典 ---")
name_score_dict = dict(zip(names, scores))
for name, score in name_score_dict.items():
    print(f"{name}: {score}")

# 2. enumerate() 输出序号
print("\n--- 任务2: enumerate ---")
for idx, (name, score) in enumerate(name_score_dict.items(), start=1):
    print(f"序号 {idx}: {name} - {score}")

# 3. 列表推导式
print("\n--- 任务3: 列表推导式 ---")
passed_names = [name for name, score in name_score_dict.items() if score >= 60]
squared_scores = [score ** 2 for score in scores]
print(f"及格学生姓名: {passed_names}")
print(f"成绩平方列表: {squared_scores}")

# 4. 基础统计
print("\n--- 任务4: 基础统计 ---")
count = len(scores)
avg_score = sum(scores) / count
max_score = max(scores)
min_score = min(scores)
pass_rate = len(passed_names) / count

print(f"人数: {count}")
print(f"平均分: {avg_score:.2f}")
print(f"最高分: {max_score}")
print(f"最低分: {min_score}")
print(f"及格率: {pass_rate:.1%}")

# 5. 浅拷贝与直接赋值的区别
print("\n--- 任务5: 直接赋值 vs copy() ---")
demo_scores = [10, 20, 30, 40]
alias = demo_scores
copied = demo_scores.copy()

alias[0] = 999  # 修改 alias
print(f"原始列表: {demo_scores}")
print(f"alias (别名): {alias} (被修改了)")
print(f"copied (副本): {copied} (未被修改)")

# 6. 元组与 zip 长度不同
print("\n--- 任务6: 元组与长度不一的 zip ---")
student_record = ("Alice", 85)
try:
    student_record[1] = 90
except TypeError as e:
    print(f"捕获到异常: {e}。原因：元组是不可变类型。")

list_a = [1, 2, 3]
list_b = ['a', 'b']
print(f"zip长度不同: {list(zip(list_a, list_b))} (以短的为准)")
```

```python
# ==========================================
# 实验4：文件读写、异常处理与模块
# ==========================================
print("="*20, "实验4", "="*20)
import os
import pickle

# 1. 路径操作与创建目录
print("\n--- 任务1: 路径操作 ---")
current_dir = os.getcwd()
print(f"当前工作目录: {current_dir}")

data_dir = os.path.join(current_dir, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print(f"数据目录已创建/存在: {data_dir}")

# 2. 写入数据文件
print("\n--- 任务2: 写入与读取文件 ---")
file_path = os.path.join(data_dir, "students.txt")
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(file_path, "w", encoding="utf-8") as f:
    f.write(content)

# 3. 定义解析函数
def parse_student(line):
    parts = line.strip().split(',')
    if len(parts) != 2:
        raise ValueError("字段数量不正确")
    name = parts[0].strip()
    if not name:
        raise ValueError("姓名不能为空")
    try:
        score = int(parts[1].strip())
    except ValueError:
        raise ValueError("成绩必须为整数")
    if not (0 <= score <= 100):
        raise ValueError("成绩必须在0-100之间")
    return {"name": name, "score": score}

# 4. 处理文件并记录错误
print("\n--- 任务4: 处理文件并捕获异常 ---")
valid_records = []
invalid_count = 0

with open(file_path, "r", encoding="utf-8") as f:
    next(f)  # 跳过表头
    for line_num, line in enumerate(f, start=2):
        if not line.strip():
            continue
        try:
            record = parse_student(line)
            valid_records.append(record)
        except ValueError as e:
            invalid_count += 1
            print(f"第 {line_num} 行错误: {e} -> 内容: {line.strip()}")

print(f"\n有效记录数: {len(valid_records)}")
print(f"无效记录数: {invalid_count}")

# 5. 写入报告文件
print("\n--- 任务5: 写入报告文件 ---")
avg_score = sum(r["score"] for r in valid_records) / len(valid_records) if valid_records else 0
pass_count = sum(1 for r in valid_records if r["score"] >= 60)
pass_rate = pass_count / len(valid_records) if valid_records else 0

report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {len(valid_records)}\n")
    f.write(f"无效记录数: {invalid_count}\n")
    f.write(f"平均分: {avg_score:.2f}\n")
    f.write(f"及格率: {pass_rate:.1%}\n")

# 追加模式演示
with open(report_path, "a", encoding="utf-8") as f:
    f.write("--- 追加测试 ---\n")

# 6. pickle 序列化
print("\n--- 任务6: pickle 序列化 ---")
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path, "rb") as f:
    loaded_records = pickle.load(f)
print(f"pickle 恢复的数据与原数据相等: {loaded_records == valid_records}")

# 7. 模拟模块导入（在 Jupyter 中直接用函数替代）
print("\n--- 任务7: 模块调用演示 ---")
print(f"有效测试: {parse_student('Alice,85')}")
try:
    parse_student('Bob,abc')
except ValueError as e:
    print(f"无效测试捕获异常: {e}")

# 8. FileNotFoundError
print("\n--- 任务8: 文件不存在异常 ---")
try:
    with open("non_existent_file.txt", "r") as f:
        pass
except FileNotFoundError as e:
    print(f"捕获到异常: {e}。说明：文件未找到。")
```
