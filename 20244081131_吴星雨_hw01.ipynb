```python
# 1. 实现 gcd 函数
def gcd(m, n):
    # 使用欧几里得算法（辗转相除法）
    while n != 0:
        m, n = n, m % n  # 将 n 赋值给 m，将 m 除以 n 的余数赋值给 n
    return m

# 2. 评估以下数字对的 GCD
pairs = [(2018, 2019), (1200, 300), (5040, 60)]

print("--- 评估数字对的 GCD ---")
for m, n in pairs:
    result = gcd(m, n)
    print(f"gcd({m}, {n}) = {result}")
```

```python
import math

# 1. 定义 euler_limit 函数
def euler_limit(n):
    """
    根据公式 (1 + 1/n)^n 近似计算 e。
    """
    return (1 + 1 / n) ** n


# 2. 定义 euler_infinite_sum 函数
def euler_infinite_sum(n):
    """
    根据公式 (1) 的前 n 项之和近似计算 e。
    注意：n=0 返回 0，n=1 返回 1 (即 1/0!)，n=2 返回 2 (1/0! + 1/1!)...
    """
    e_approx = 0.0
    factorial = 1.0  # 用于累乘计算阶乘

    # 循环 n 次，对应公式中的前 n 项 (k 从 0 到 n-1)
    for k in range(n):
        if k > 0:
            factorial *= k  # 计算 k!
        e_approx += 1 / factorial

    return e_approx


# 3. 定义 euler_approx 函数
def euler_approx(tolerance):
    """
    使用 while 循环累加公式 (1) 的项，直到近似值与真实值 e 的误差在 tolerance 之内。
    """
    true_e = math.exp(1)
    e_approx = 0.0
    factorial = 1.0
    k = 0

    # 当误差大于容差时，继续累加
    while abs(e_approx - true_e) > tolerance:
        if k > 0:
            factorial *= k
        e_approx += 1 / factorial
        k += 1

    return e_approx


# 4. 定义 print_euler_sum_table 函数
def print_euler_sum_table(n):
    """
    打印从 euler_infinite_sum(1) 到 euler_infinite_sum(n) 的值，每行一个。
    """
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

if __name__ == "__main__":
    print("=== 测试 1: euler_limit ===")
    print(f"euler_limit(10000): {euler_limit(10000)}")
    print(f"真实 e 值: {math.exp(1)}")

    print("\n=== 测试 2: euler_infinite_sum ===")
    print(f"euler_infinite_sum(0): {euler_infinite_sum(0)}") # 应返回 0
    print(f"euler_infinite_sum(1): {euler_infinite_sum(1)}") # 应返回 1
    print(f"euler_infinite_sum(4): {euler_infinite_sum(4)}") # 应返回 2.6666...
    print(f"euler_infinite_sum(10): {euler_infinite_sum(10)}")

    print("\n=== 测试 3: euler_approx ===")
    print(f"euler_approx(0.01): {euler_approx(0.01)}")
    print(f"euler_approx(0.00001): {euler_approx(0.00001)}")

    print("\n=== 测试 4: print_euler_sum_table ===")
    print_euler_sum_table(5)
```

```python
def is_power_of_2(n):

    if n == 1:
        return True

    # 如果 n 小于 1，或者不能被 2 整除，则不是 2 的幂
    # 注意：题目假设输入是正整数，但为了逻辑严谨，我们加上 n < 1 的判断
    if n < 1 or n % 2 != 0:
        return False

    # 不断将 n 除以 2，直到不能再整除为止
    while n > 1:
        if n % 2 != 0:  # 如果在除以 2 的过程中出现了奇数，说明不是 2 的幂
            return False
        n = n // 2      # 使用整数除法

    return True


# 2. 推广到 is_power 函数
def is_power(b, n):
    """
    判断正整数 n 是否是 b 的幂。
    接收两个正整数作为参数：b (底数) 和 n (待判断的数)。
    """
    # 任何正整数的 0 次幂都是 1，所以如果 n 是 1，它总是任何 b 的幂 (b^0 = 1)
    if n == 1:
        return True

    # 如果底数 b 是 1，那么 1 的任何次幂都是 1。
    # 由于前面已经判断 n != 1，所以此时 n 不可能是 1 的幂
    if b == 1:
        return False

    # 如果 n 小于 b，且不是 1，那么它不可能是 b 的幂 (正整数次幂)
    if n < b:
        return False

    # 不断将 n 除以 b，直到不能再整除为止
    while n > 1:
        if n % b != 0:  # 如果出现了不能整除的情况，说明不是 b 的幂
            return False
        n = n // b      # 使用整数除法

    return True


# --- 以下是测试代码，用于验证函数是否正常工作 ---

if __name__ == "__main__":
    print("=== 测试 is_power_of_2 ===")
    # 测试 2 的幂
    for num in [1, 2, 4, 8, 16, 32, 64, 1024]:
        print(f"is_power_of_2({num}) = {is_power_of_2(num)}")

    # 测试非 2 的幂
    for num in [3, 5, 6, 7, 12, 100, 2018]:
        print(f"is_power_of_2({num}) = {is_power_of_2(num)}")

    print("\n=== 测试 is_power ===")
    # 测试 3 的幂
    print(f"is_power(3, 1) = {is_power(3, 1)}")
    print(f"is_power(3, 3) = {is_power(3, 3)}")
    print(f"is_power(3, 9) = {is_power(3, 9)}")
    print(f"is_power(3, 27) = {is_power(3, 27)}")
    print(f"is_power(3, 12) = {is_power(3, 12)}")

    # 测试底数为 1 的情况
    print(f"is_power(1, 1) = {is_power(1, 1)}")
    print(f"is_power(1, 5) = {is_power(1, 5)}")
```
