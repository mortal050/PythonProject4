```python
class Vector:
    def __init__(self, dim, elements=None):
        # 检查维度类型
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError("维度必须是整数")
        # 检查维度非负
        if dim < 0:
            raise ValueError("维度不能为负数")

        # 处理未提供元素的情况（默认创建全 0 向量）
        if elements is None:
            elements = [0] * dim

        # 检查元素类型
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须由列表或元组构成")

        # 检查维度与元素个数是否一致
        if len(elements) != dim:
            raise ValueError("维度与元素个数不一致")

        # 检查内部元素是否全为整数或浮点数
        for x in elements:
            if not isinstance(x, (int, float)) or isinstance(x, bool):
                raise TypeError("元素必须全都是整数或浮点数")

        # 设置为实例属性
        self.dim = dim
        self.elements = list(elements)  # 转换为列表，便于后续操作

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    # 维度检查辅助函数
    def _check_dim(self, other):
        if not isinstance(other, Vector):
            raise TypeError("操作对象必须是 Vector 类型")
        if self.dim != other.dim:
            raise ValueError("两个向量的维度不一致")

    # 3. 比较运算符
    def __eq__(self, other):
        self._check_dim(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self._check_dim(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check_dim(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_dim(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_dim(other)
        return self.elements >= other.elements

    # 4. 内积方法
    def dot(self, other):
        self._check_dim(other)
        return float(sum(x * y for x, y in zip(self.elements, other.elements)))

    # 5. 乘法运算符
    def __mul__(self, other):
        # 情况 A：与标量相乘 (如 v * 2)
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return Vector(self.dim, [x * other for x in self.elements])

        # 情况 B：与另一个 Vector 相乘 (如 v * w)
        elif isinstance(other, Vector):
            self._check_dim(other)
            return Vector(self.dim, [x * y for x, y in zip(self.elements, other.elements)])

        else:
            raise TypeError("乘法操作仅支持 标量 (int/float) 或 Vector")

    # 支持标量左乘 (如 2 * v)
    def __rmul__(self, other):
        return self.__mul__(other)


# ================= 测试代码 =================
if __name__ == "__main__":
    v1 = Vector(2, (2, 4))
    v2 = Vector(2, (2, 5))
    v3 = Vector(2, (4, 5))
    v4 = Vector(3, (1, 2, 3))

    print(v1)
    print("相等:", v1 == Vector(2, (2, 4)))
    print("小于:", v1 < v2)
    print("大于等于:", v2 >= v1)

    print("内积:", v1.dot(v2))

    print("标量右乘:", v1 * 2)
    print("标量左乘:", 2 * v1)
    print("向量相乘:", Vector(2, (2, 3)) * Vector(2, (4, 5)))

    try:
        print(v1 < v4)
    except ValueError as e:
        print("异常:", e)
```

```python
def is_palindrome(s):
    """
    1. 判断是否为回文（忽略空格和大小写）
    """
    # 1. 转换为小写
    # 2. 去除所有空格
    cleaned_s = s.lower().replace(" ", "")
    # 3. 比较清理后的字符串与它的反转
    return cleaned_s == cleaned_s[::-1]


def is_abecedarian(s):
    """
    2. 判断是否按字母顺序排列（忽略空格，允许字母重复）
    """
    # 1. 去除所有空格
    # 2. 转换为小写（避免大小写带来的比较偏差）
    cleaned_s = s.lower().replace(" ", "")

    # 逐个比较相邻的字符
    for i in range(len(cleaned_s) - 1):
        if cleaned_s[i] > cleaned_s[i+1]:
            return False
    return True


def remove_vowels(s):
    """
    3. 去掉字符串中所有的元音（a, e, i, o, u）
    """
    vowels = "aeiouAEIOU"  # 为了严谨，同时处理了大写元音
    result = ""
    for char in s:
        if char not in vowels:
            result += char
    return result


# ================= 测试代码 =================
if __name__ == "__main__":
    print("--- 测试 1: is_palindrome ---")
    print(f"tacocat: {is_palindrome('tacocat')}")
    print(f"T A C O cat: {is_palindrome('T A C O cat')}")
    print(f"hello: {is_palindrome('hello')}")

    print("\n--- 测试 2: is_abecedarian ---")
    print(f"adder: {is_abecedarian('adder')}")
    print(f"beet: {is_abecedarian('beet')}")
    print(f"abcd efgh xyz: {is_abecedarian('abcd efgh xyz')}")
    print(f"dog: {is_abecedarian('dog')}")
    print(f"cat: {is_abecedarian('cat')}")

    print("\n--- 测试 3: remove_vowels ---")
    print(f"cat: {remove_vowels('cat')}")
    print(f"audio: {remove_vowels('audio')}")
    print(f"aeiou: {remove_vowels('aeiou')}")
```

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("参数 t 必须是元组 (tuple)")
    if not all(isinstance(x, (int, float)) and not isinstance(x, bool) for x in t):
        raise TypeError("元组 t 中的元素必须全都是整数或浮点数")
    if not isinstance(s, (int, float)) or isinstance(s, bool):
        raise TypeError("标量 s 必须是整数或浮点数")
    return tuple(x * s for x in t)


def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("参数 x 和 y 都必须是元组 (tuple)")
    if len(x) != len(y):
        raise ValueError("两个向量的长度必须一致")
    if not all(isinstance(i, (int, float)) for i in x) or not all(isinstance(i, (int, float)) for i in y):
        raise TypeError("向量中的元素必须全都是整数或浮点数")
    return float(sum(xi * yi for xi, yi in zip(x, y)))


def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    for row in mx:
        if not isinstance(row, tuple):
            return False
    row_length = len(mx[0])
    for row in mx:
        if len(row) != row_length:
            return False
    for row in mx:
        for item in row:
            if not isinstance(item, (int, float)) or isinstance(item, bool):
                return False
    return True


def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise ValueError("输入的矩阵不合法（请检查数据结构、维度或元素类型）")
    if not isinstance(v, tuple):
        raise TypeError("向量 v 必须是元组")
    if not all(isinstance(i, (int, float)) and not isinstance(i, bool) for i in v):
        raise TypeError("向量 v 中的元素必须全都是整数或浮点数")
    if len(mx) > 0 and len(mx[0]) != len(v):
        raise ValueError("矩阵的列数必须等于向量的长度")
    result = []
    for row in mx:
        dot_product = sum(mx_element * v_element for mx_element, v_element in zip(row, v))
        result.append(dot_product)
    return tuple(result)


if __name__ == "__main__":
    print(vec_scalar_mult((1, 2, 3), 2.5))
    try:
        vec_scalar_mult([1, 2], 2)
    except TypeError as e:
        print("TypeError:", e)

    print(vec_inner_product((1, 2, 3), (4, 5, 6)))
    try:
        vec_inner_product((1, 2), (1, 2, 3))
    except ValueError as e:
        print("ValueError:", e)

    valid_mx = ((1, 2), (3, 4), (5, 6))
    invalid_mx = ((1, 2), (3, 'a'))
    print(check_valid_mx(valid_mx))
    print(check_valid_mx(invalid_mx))
    print(check_valid_mx((1, 2, 3)))

    print(mx_vec_mult(((1, 2), (3, 4)), (5, 6)))
    try:
        mx_vec_mult(((1, 2), (3, 4)), (5, 6, 7))
    except ValueError as e:
        print("ValueError:", e)
```
