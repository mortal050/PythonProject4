```python
import numpy as np

print("="*50)
print("1. 分别创建整数、浮点数和布尔值 ndarray")
print("="*50)
arr_int = np.array([1, 2, 3])
arr_float = np.array([1.0, 2.0, 3.0])
arr_bool = np.array([True, False, True])

print(f"整数数组 - dtype: {arr_int.dtype}, shape: {arr_int.shape}, ndim: {arr_int.ndim}")
print(f"浮点数组 - dtype: {arr_float.dtype}, shape: {arr_float.shape}, ndim: {arr_float.ndim}")
print(f"布尔数组 - dtype: {arr_bool.dtype}, shape: {arr_bool.shape}, ndim: {arr_bool.ndim}")
print("\n[解释] NumPy数组与Python列表的区别：")
print("Python 列表可以包含不同类型的数据，而 NumPy 数组要求所有元素类型必须一致（同质化）。")
print("NumPy 底层由 C 语言实现，内存连续，支持向量化运算，因此在处理大规模数据时，速度比 Python 列表快得多，且占用内存更小。")


print("\n" + "="*50)
print("2. 用 np.arange(12) 创建一维数组并改变形状")
print("="*50)
arr_1d = np.arange(12)
print("原始一维数组:\n", arr_1d)

arr_3x4 = arr_1d.reshape((3, 4))
print("\n变为 (3, 4) 形状:\n", arr_3x4)

arr_2x3x2 = arr_1d.reshape((2, 3, 2))
print("\n变为 (2, 3, 2) 形状:\n", arr_2x3x2)

print("\n[解释] 改变形状时元素总数必须满足的条件：")
print("新形状的各维度乘积（即元素总数）必须与原数组的元素总数完全相等。例如 12 个元素不能变为 (3, 5)。")


print("\n" + "="*50)
print("3. 对形状为 (3, 4) 的数组完成索引操作")
print("="*50)
# 取第 2 行 (索引为 1)
row_2 = arr_3x4[1, :]
print("取第 2 行:\n", row_2, "\n结果 shape:", row_2.shape)

# 取第 1 至第 3 列 (索引 0 到 2，左闭右开)
cols_1_3 = arr_3x4[:, 0:3]
print("\n取第 1 至第 3 列:\n", cols_1_3, "\n结果 shape:", cols_1_3.shape)

# 取大于 6 的元素
gt_6 = arr_3x4[arr_3x4 > 6]
print("\n取大于 6 的元素:\n", gt_6, "\n结果 shape:", gt_6.shape)


print("\n" + "="*50)
print("4. 使用 np.arange(10) 进行逻辑筛选")
print("="*50)
arr_10 = np.arange(10)
print("原始数组:", arr_10)

# 小于 2 或大于 7
# 注意：使用 | 和 & 时，括号必不可少
res_or_1 = arr_10[(arr_10 < 2) | (arr_10 > 7)]
res_or_2 = arr_10[np.logical_or(arr_10 < 2, arr_10 > 7)]
print("\n选出'小于 2 或大于 7'的元素:")
print("使用括号+|:", res_or_1)
print("使用 logical_or:", res_or_2)

# 大于 3 且小于 7
res_and_1 = arr_10[(arr_10 > 3) & (arr_10 < 7)]
res_and_2 = arr_10[np.logical_and(arr_10 > 3, arr_10 < 7)]
print("\n选出'大于 3 且小于 7'的元素:")
print("使用括号+&:", res_and_1)
print("使用 logical_and:", res_and_2)


print("\n" + "="*50)
print("5. 构造两个 (2, 2) 数组，对比逐元素乘法与矩阵乘法")
print("="*50)
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])
print("数组 A:\n", A)
print("数组 B:\n", B)

print("\n--- 逐元素乘法 ---")
print("使用 * 运算符:\n", A * B)
print("使用 np.multiply():\n", np.multiply(A, B))

print("\n--- 矩阵乘法 ---")
print("使用 @ 运算符:\n", A @ B)
print("使用 np.matmul():\n", np.matmul(A, B))

print("\n[解释] 逐元素乘法与矩阵乘法的区别：")
print("1. 逐元素乘法 (* 或 np.multiply)：对应位置的元素直接相乘，即 C[i][j] = A[i][j] * B[i][j]。")
print("2. 矩阵乘法 (@ 或 np.matmul)：符合线性代数规则，结果矩阵第 i 行第 j 列的元素，是 A 的第 i 行与 B 的第 j 列的点积。")
```

```python
import numpy as np
import pandas as pd # 第5题生成表格变量会用到

print("="*50)
print("1. 生成模拟成绩、截断并输出统计信息")
print("="*50)
# 设定随机种子以保证结果可复现（可选）
np.random.seed(42)

# 生成 1000 个均值 70，标准差 10 的成绩
scores = np.random.normal(loc=70, scale=10, size=1000)
# 用 np.clip 限制在 0~100 范围
scores_clipped = np.clip(scores, 0, 100)

print(f"样本数: {scores_clipped.size}")
print(f"最小值: {scores_clipped.min():.2f}")
print(f"最大值: {scores_clipped.max():.2f}")
print(f"均值: {scores_clipped.mean():.2f}")
print(f"标准差: {scores_clipped.std():.2f}")
print(f"极差 (Max - Min): {scores_clipped.max() - scores_clipped.min():.2f}")


print("\n" + "="*50)
print("2. 不放回抽样：均匀抽样 vs 按整数大小成比例加权抽样")
print("="*50)
arr = np.arange(1, 11) # [1, 2, 3, ..., 10]
print("原始数组:", arr)

# 均匀抽样：每个元素被抽中的概率相同
# replace=False 表示不放回
uniform_sample = np.random.choice(arr, size=5, replace=False)
print("\n[均匀抽样] 抽出 5 个整数:", uniform_sample)

# 按权重抽样：数字本身作为权重（数字越大被抽中概率越高）
# p 参数要求所有概率和为 1
probabilities = arr / np.sum(arr)
weighted_sample = np.random.choice(arr, size=5, replace=False, p=probabilities)
print("[加权抽样] 抽出 5 个整数:", weighted_sample)


print("\n" + "="*50)
print("3. 随机排列：保留原数组 vs 原地打乱")
print("="*50)
original_arr = np.arange(1, 11)
print("操作前的原数组:", original_arr)

# 方式一：保留原数组（返回一个新的打乱后的数组，原数组不变）
arr_shuffled_new = np.random.permutation(original_arr)
print("\n使用 np.random.permutation() 后：")
print("新的打乱数组:", arr_shuffled_new)
print("原数组（未变）:", original_arr)

# 方式二：原地打乱（直接在原数组上操作，原数组会改变）
arr_shuffled_inplace = original_arr.copy() # 先复制一份，以免影响上面的演示
np.random.shuffle(arr_shuffled_inplace)
print("\n使用 np.random.shuffle() 后：")
print("原数组已变为:", arr_shuffled_inplace)
print("[说明] shuffle 是原地操作(in-place)，permutation 返回新数组。")


print("\n" + "="*50)
print("4. 缺失值处理：普通均值 vs 忽略缺失值")
print("="*50)
# 为了不覆盖第一题的数据，这里复制一份来操作
scores_nan = scores_clipped.copy()

# 将第 10, 20, 30 个值设为 np.nan (索引是 9, 19, 29)
scores_nan[[9, 19, 29]] = np.nan

# 普通均值（遇到 nan 会传染，结果变成 nan）
normal_mean = scores_nan.mean()
# 忽略缺失值后的均值
nan_mean = np.nanmean(scores_nan)
# 统计缺失值个数
nan_count = np.isnan(scores_nan).sum()

print(f"包含缺失值的数组前 35 个元素:\n{scores_nan[:35]}")
print(f"\n普通均值 (mean): {normal_mean}")
print(f"忽略缺失值后的均值 (nanmean): {nan_mean:.2f}")
print(f"缺失值个数: {nan_count}")
print("\n[说明] nan 对普通统计函数的影响：")
print("np.nan 在计算时具有'传染性'。如果不使用专门的忽略缺失值函数（如 np.nanmean），")
print("任何包含 np.nan 的数组在进行普通求和、求均值等操作时，结果都会变成 np.nan。")


print("\n" + "="*50)
print("5. 整理成包含缺失值和序号的表格变量 (为实验4准备)")
print("="*50)
# 创建序号数组 (1~1000)
indices = np.arange(1, 1001)

# 组织成字典，生成 DataFrame 表格
df = pd.DataFrame({
    '序号': indices,
    '成绩': scores_nan
})

print("表格变量 'df' 已创建，预览前 30 行 (包含第10,20,30行的 NaN)：")
print(df.head(30))
print("\n该表格已保存在内存变量 df 中，供实验4直接调用。")
```

```python
import numpy as np
import scipy.linalg as la
import scipy.optimize as opt
import scipy.stats as stats
import matplotlib.pyplot as plt

# 解决 Matplotlib 中文显示问题
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei'] # 优先使用黑体
plt.rcParams['axes.unicode_minus'] = False # 正常显示负号


print("="*50)
print("1. 使用 scipy.linalg.solve() 解线性方程组")
print("="*50)
# 方程组: 3x + y = 9, x + 2y = 8
# 系数矩阵 A 和常数项向量 b
A = np.array([[3, 1],
              [1, 2]])
b = np.array([9, 8])

# 求解
solution = la.solve(A, b)
print(f"求解结果: x = {solution[0]}, y = {solution[1]}")

# 验证：A * x = b
verification = np.dot(A, solution)
print(f"代入验证 A @ x 结果: {verification}")
print(f"与原始 b 是否一致: {np.allclose(verification, b)}")


print("\n" + "="*50)
print("2. 使用 scipy.optimize.root() 求方程 x^2 - 2 = 0 的根")
print("="*50)
# 定义目标函数
def func(x):
    return x**2 - 2

# 求解正根（初始猜测值设为 1）
res_pos = opt.root(func, x0=1.0)
# 求解负根（初始猜测值设为 -1）
res_neg = opt.root(func, x0=-1.0)

print(f"正根求解: 是否成功={res_pos.success}, 根={res_pos.x[0]:.4f}, 函数值={func(res_pos.x[0]):.2e}")
print(f"负根求解: 是否成功={res_neg.success}, 根={res_neg.x[0]:.4f}, 函数值={func(res_neg.x[0]):.2e}")
print("[说明] 函数值极接近 0，说明求得的解是准确的。")


print("\n" + "="*50)
print("3. 使用 scipy.optimize.minimize() 求极小值")
print("="*50)
# 定义函数 f(x) = x^2 + x + 2
def f(x):
    return x**2 + x + 2

# 求极小值
res_min = opt.minimize(f, x0=0.0) # 初始猜测设为 0

print(f"优化是否成功: {res_min.success}")
print(f"极小值点 x = {res_min.x[0]:.4f}")
print(f"极小值 f(x) = {res_min.fun:.4f}")
print("[比较] 解析解: 导数为 2x + 1 = 0 => x = -0.5，代入得 f(-0.5) = 1.75。结果与解析解一致。")


print("\n" + "="*50)
print("4. 独立样本 t 检验")
print("="*50)
# 固定随机种子
np.random.seed(42)

# 生成两组样本
group1 = np.random.normal(loc=0, scale=1, size=100)
group2 = np.random.normal(loc=1, scale=1, size=100)

# 独立样本 t 检验
t_stat, p_value = stats.ttest_ind(group1, group2)

print(f"t 统计量: {t_stat:.4f}")
print(f"p 值: {p_value:.4e}")

if p_value < 0.05:
    print("判断: p 值 < 0.05，拒绝原假设。两组数据存在显著差异。")
else:
    print("判断: p 值 >= 0.05，不能拒绝原假设。两组数据不存在显著差异。")


print("\n" + "="*50)
print("5. Matplotlib 数据可视化")
print("="*50)
# 注意：这里假设上方已经运行过实验2的代码，存在 scores_clipped 变量
# 如果没有，我在这里为你重新生成一下（提取有效成绩，无 NaN）
if 'scores_clipped' not in locals():
    print("未检测到实验2的成绩变量，自动为您重新生成...")
    np.random.seed(42)
    raw_scores = np.random.normal(loc=70, scale=10, size=1000)
    scores_clipped = np.clip(raw_scores, 0, 100)

# 创建 1 行 2 列的画布
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 画直方图
axes[0].hist(scores_clipped, bins=20, color='skyblue', edgecolor='black', alpha=0.7)
axes[0].set_title("模拟成绩分布直方图")
axes[0].set_xlabel("成绩")
axes[0].set_ylabel("频数")

# 画箱线图
axes[1].boxplot(scores_clipped, patch_artist=True,
                boxprops=dict(facecolor='lightgreen'))
axes[1].set_title("模拟成绩箱线图")
axes[1].set_ylabel("成绩")
axes[1].set_xticks([1]) # 只显示一个刻度
axes[1].set_xticklabels(["成绩"])

plt.tight_layout()
plt.show() # 直接在 Notebook 中显示


# 绘制函数折线图
# 生成 x 数据 (0 到 5 之间，100 个点)
x_vals = np.linspace(0, 5, 100)

plt.figure(figsize=(8, 5))
plt.plot(x_vals, x_vals, label='y = x', color='blue')
plt.plot(x_vals, x_vals**2, label='y = x^2', color='orange')
plt.plot(x_vals, x_vals**3, label='y = x^3', color='green')

plt.title("函数 y = x, y = x^2, y = x^3 的折线图")
plt.xlabel("x 轴")
plt.ylabel("y 轴")
plt.legend() # 显示图例
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import pandas as pd
import numpy as np

# 解决 DataFrame 打印时部分列被折叠隐藏的问题
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)


print("="*60)
print("1. 创建 DataFrame 并计算总分、平均分")
print("="*60)
records = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82],
}

# 创建 DataFrame 并将 Name 设置为行索引
df = pd.DataFrame(records).set_index('Name')

# 计算 Total 和 Average (只对三科成绩列进行计算)
subjects = ['Math', 'English', 'Python']
df['Total'] = df[subjects].sum(axis=1)
df['Average'] = df[subjects].mean(axis=1)

print("行索引 (Index):", df.index.tolist())
print("列名 (Columns):", df.columns.tolist())
print("\n前 3 行数据:")
print(df.head(3))
print("\n基本统计信息:")
print(df.describe())


print("\n" + "="*60)
print("2. 使用 loc 和 iloc 读取数据及筛选")
print("="*60)
# 使用 loc (基于标签索引)
print("使用 loc 读取 Carol 的记录:")
print(df.loc['Carol'])

# 使用 iloc (基于位置索引，Carol 在第 3 行，索引为 2)
print("\n使用 iloc 读取 Carol 的记录:")
print(df.iloc[2])

# 筛选平均分不低于 80 的学生
print("\n平均分不低于 80 的学生:")
print(df[df['Average'] >= 80])


print("\n" + "="*60)
print("3. 计算列均值和行均值")
print("="*60)
# 列均值 (按列求平均，即每科的平均分)
print("三科成绩的列均值 (mean(axis=0)):")
print(df[subjects].mean(axis=0))

# 行均值 (按行求平均，即每名学生的平均分)
# 这里直接验证我们之前计算的 Average 列
print("\n每名学生的行均值 (mean(axis=1)):")
print(df[subjects].mean(axis=1))


print("\n" + "="*60)
print("4. 使用 apply 和 agg 及 rank")
print("="*60)
# 使用 apply 计算每名学生的三科最高分
df['Max_Score'] = df[subjects].apply(np.max, axis=1)
print("新增每位学生的最高分 (apply):")
print(df[['Math', 'English', 'Python', 'Max_Score']])

# 使用 agg 一次性计算每科的最小值、均值和最大值
print("\n每科的最小值、均值、最大值 (agg):")
print(df[subjects].agg(['min', 'mean', 'max']))

# 任选一列排名 (例如 Math，method='min' 表示并列时取最小名次)
df['Math_Rank'] = df['Math'].rank(ascending=False, method='min')
print("\n数学成绩排名 (rank，分数越高排名越靠前):")
print(df[['Math', 'Math_Rank']])


print("\n" + "="*60)
print("5. 按 Group 分组统计及 transform")
print("="*60)
# 按 Group 分组统计人数和三科平均分
print("按 Group 分组统计:")
group_stats = df.groupby('Group').agg(
    人数=('Math', 'count'),
    Math平均=('Math', 'mean'),
    English平均=('English', 'mean'),
    Python平均=('Python', 'mean')
)
print(group_stats)

# 用 transform('mean') 为每行增加其所在组的 Python 平均分
df['Group_Python_Mean'] = df.groupby('Group')['Python'].transform('mean')
print("\n增加所在组 Python 均分 (transform):")
print(df[['Group', 'Python', 'Group_Python_Mean']])


print("\n" + "="*60)
print("6. 使用 filter 筛选组")
print("="*60)
# 保留 Python 平均分不低于 80 的组
# 这里结合 lambda 函数，判断组的均值是否 >= 80
filtered_df = df.groupby('Group').filter(lambda x: x['Python'].mean() >= 80)
print("保留 Python 均分 >= 80 的组及其学生:")
print(filtered_df)


print("\n" + "="*60)
print("7. 协方差与皮尔逊相关系数")
print("="*60)
# 计算协方差矩阵
cov_matrix = df[subjects].cov()
print("三科成绩的协方差矩阵:")
print(cov_matrix)

# 计算皮尔逊相关系数矩阵
corr_matrix = df[subjects].corr(method='pearson')
print("\n三科成绩的皮尔逊相关系数矩阵:")
print(corr_matrix)

print("\n[解释] 我们以 Math(数学) 和 Python 为例：")
print(f"Math 与 Python 的协方差为: {cov_matrix.loc['Math', 'Python']:.2f}。")
print("协方差的符号（正号）表示两者呈正相关，即数学成绩越高，Python 成绩往往也越高。")
print(f"Math 与 Python 的相关系数为: {corr_matrix.loc['Math', 'Python']:.2f}。")
```
